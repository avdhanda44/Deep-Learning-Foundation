# Convolution by Hand

Calculate a valid 2D convolution-style operation with a tiny image and filter, then verify every output with PyTorch.

```python
import torch
from torch import nn

torch.set_num_threads(1)
dtype = torch.float64
print('CPU | hand convolution check')
```

```python
image = torch.tensor([[[[0., 1., 0.], [0., 1., 0.], [0., 1., 0.]]]], dtype=dtype)
kernel = torch.tensor([[0., 1.], [0., 1.]], dtype=dtype)
manual = torch.tensor([[[[2., 0.], [2., 0.]]]], dtype=dtype)
layer = nn.Conv2d(1, 1, kernel_size=2, bias=False).double()
with torch.no_grad():
    layer.weight.copy_(kernel.view(1, 1, 2, 2))
calculated = layer(image)
torch.testing.assert_close(calculated, manual)
assert calculated.shape == (1, 1, 2, 2)
print('manual and PyTorch output:')
print(calculated[0, 0])
```

```python
pooled = nn.MaxPool2d(2)(calculated)
torch.testing.assert_close(pooled, torch.tensor([[[[2.]]]], dtype=dtype))
print('max pooled:', pooled.item())
print('parameters:', sum(parameter.numel() for parameter in layer.parameters()))
```
