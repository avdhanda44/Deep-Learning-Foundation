# Generative Modeling Foundations

Sample from a known two-component distribution and compare empirical samples with its intended mixture proportions. This introduces sampling, not a trained generative model.

```python
import torch

torch.manual_seed(37)
torch.set_num_threads(1)
print('CPU | sampling a synthetic mixture')
```

```python
count = 10000
component = torch.bernoulli(torch.full((count,), 0.7)).long()
centers = torch.tensor([-2., 2.])
samples = centers[component] + 0.5 * torch.randn(count)
fraction_first = (component == 0).float().mean().item()
means = torch.stack([samples[component == label].mean() for label in (0, 1)])
assert abs(fraction_first - 0.3) < 0.03
assert torch.allclose(means, centers, atol=0.08)
print(f'component probabilities: 0.30 / 0.70 | observed first: {fraction_first:.3f}')
print('sampled component means:', means.tolist())
```

Here the sampling rule is known and fixed. A learned generative model has to estimate a useful sampling procedure from data; matching a few summary statistics is not enough to prove that it captured the full data distribution.
