# Transformer Encoder and Decoder

Run a tiny encoder-decoder Transformer on a source sequence and a shifted target prefix. Inspect source, target, and output shapes and apply a causal target mask.

```python
import torch
from torch import nn

torch.manual_seed(30)
torch.set_num_threads(1)
print('CPU | encoder-decoder shapes and masking')
```

```python
model = nn.Transformer(d_model=8, nhead=2, num_encoder_layers=1, num_decoder_layers=1,
                       dim_feedforward=16, dropout=0.0, batch_first=True)
source = torch.randn(1, 3, 8)
target_prefix = torch.randn(1, 2, 8)
target_length = target_prefix.shape[1]
causal_mask = torch.triu(torch.ones(target_length, target_length, dtype=torch.bool), diagonal=1)
output = model(source, target_prefix, tgt_mask=causal_mask)
assert output.shape == (1, 2, 8)
print('source:', tuple(source.shape), 'target prefix:', tuple(target_prefix.shape))
print('decoder output:', tuple(output.shape))
print('causal target mask:')
print(causal_mask.int())
```

The target prefix is what the decoder has seen so far; the source is available through cross-attention. Real training data also needs aligned decoder inputs and labels shifted by one token, plus padding masks for unequal lengths.
